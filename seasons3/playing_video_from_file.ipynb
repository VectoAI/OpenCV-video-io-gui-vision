```python
import cv2 as cv 
import matplotlib.pyplot as plt 
import numpy as np 
import os 
```

```python
# dynamic path 
current_path = os.getcwd()
project_dir = os.path.dirname(current_path)
video_path = os.path.join(project_dir, 'videos/race_car.mp4')
```

```python
# reading video from a file and show 
cap = cv.VideoCapture(video_path)
ret = True

while ret : 
    ret, frame = cap.read()
    if not ret : 
        print('exiting ... ')
        break
    gray = cv.cvtColor(frame, cv.COLOR_BGR2GRAY)
    cv.imshow('video', frame)
    cv.imshow('gray', gray)
    if cv.waitKey(1) & 0XFF == 27: 
        break
    
cap.release()
cv.destroyAllWindows()
```
