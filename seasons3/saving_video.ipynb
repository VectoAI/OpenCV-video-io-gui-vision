```python
import numpy as np 
import matplotlib.pyplot as plt 
import cv2 as cv 
import os 
```

```python
# dynamic path 
current_path = os.getcwd()
project_dir = os.path.dirname(current_path)
webcam_output_path = os.path.join(project_dir, 'videos/output.avi')
video_output_path = os.path.join(project_dir, 'videos/race_car_gray.avi')
video_input_path = os.path.join(project_dir, 'videos/race_car.mp4')
```

```python
cap = cv.VideoCapture(0)
fourcc = cv.VideoWriter_fourcc(*'XVID')
ret, frame = cap.read()

out = cv.VideoWriter(webcam_output_path, fourcc, 20.0, (frame.shape[0], frame.shape[1]))
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print("Can't receive frame (stream end?). Exiting ...")
        break
    frame = cv.flip(frame, 1)
    # write the flipped frame
    out.write(frame)
    cv.imshow('frame', frame)
    if cv.waitKey(1) == ord('q'):
        break
# Release everything if job is finished
cap.release()
out.release()
cv.destroyAllWindows()
```

```python
# Load video clip
cap = cv.VideoCapture(video_input_path)

# Get the height and width of the frame (required to be an interger)
w = int(cap.get(cv.CAP_PROP_FRAME_WIDTH)) 
h = int(cap.get(cv.CAP_PROP_FRAME_HEIGHT))

# Define the codec and create VideoWriter object. The output is stored in '*.avi' file.
out = cv.VideoWriter(video_output_path, cv.VideoWriter_fourcc('M','J','P','G'), cap.get(cv.CAP_PROP_FPS), (w, h), False)

while True:

    ret, frame = cap.read()

    if ret == True:
        gray = cv.cvtColor(frame, cv.COLOR_BGR2GRAY)
        out.write(gray)
    else:
        print("Done")
        break
        
cap.release()
out.release()
cv.destroyAllWindows()
```
