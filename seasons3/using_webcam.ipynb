```python
import cv2 as cv 
import matplotlib.pyplot as plt 
import numpy as np 
```

```python
# reading image from webcam 
cap = cv.VideoCapture(0)

while True : 
    ret, frame = cap.read()
    flipped = cv.flip(frame, 1)
    image = np.zeros(frame.shape[:2], np.uint8)
    b, g, r = cv.split(flipped)
    image_green = cv.merge((image, g, image))
    image_red = cv.merge((image, image, r))
    image_blue = cv.merge((b, image, image))
    cv.imshow('webcam_green', image_green)
    cv.imshow('webcam_red', image_red)
    cv.imshow('webcam_blue', image_blue)
    
    if cv.waitKey(1) & 0xFF == 27:
        break
        
# Release camera and close windows
cap.release()
cv.destroyAllWindows() 
plt.figure(figsize=(6, 6))
plt.subplot(221)
plt.imshow(flipped[..., ::-1])
plt.subplot(222)
plt.imshow(image_green[..., ::-1])
plt.subplot(223)
plt.imshow(image_red[..., ::-1])
plt.subplot(224)
plt.imshow(image_blue[..., ::-1])
```
